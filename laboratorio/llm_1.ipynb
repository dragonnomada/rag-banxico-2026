## Inicializar un modelo local en forma de servidor con Llama

```bash
llama-server \
    -m Qwen2.5-3B-Instruct-Q4_K_M.gguf \
    --host 0.0.0.0 \
    --port 8080 \
    -c 4096 \
    -ngl 99
```

````python
from openai import OpenAI


cliente = OpenAI(
    base_url="http://localhost:8080/v1",
    api_key="local"
)


respuesta = cliente.chat.completions.create(
    model="local",
    messages=[
        {
            "role": "system",
            "content": "Eres un asistente especializado en análisis financiero."
        },
        {
            "role": "user",
            "content": """
                Explica en términos sencillos qué significa inflación
                y proporciona un ejemplo relacionado con el precio de alimentos.
            """
        }
    ],
    temperature=0.7
)


print(respuesta.choices[0].message.content)
````
