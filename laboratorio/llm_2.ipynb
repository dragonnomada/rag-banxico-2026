# Modelo LLM local con JSON estructurado

```python
from openai import OpenAI
```

```python
cliente = OpenAI(
    base_url="http://localhost:8080/v1",
    api_key="local"
)
```

```python
texto = """
TRANSFERENCIA SPEI
Fecha: 23/09/2026
Importe: $15,250.00
Beneficiario: Empresa Ejemplo SA de CV
Concepto: Pago factura 382
Referencia: 829102
"""
```

```python
respuesta = cliente.chat.completions.create(
    model="local",
    messages=[
        {
            "role": "system",
            "content": "Extrae información estructurada de documentos financieros."
        },
        {
            "role": "user",
            "content": texto
        }
    ],
    response_format={
        "type": "json_schema",
        "schema": {
            "type": "object",
            "properties": {
                "fecha": {
                    "type": "string"
                },
                "importe": {
                    "type": "number"
                },
                "beneficiario": {
                    "type": "string"
                },
                "concepto": {
                    "type": "string"
                },
                "referencia": {
                    "type": "string"
                }
            },
            "required": [
                "fecha",
                "importe",
                "beneficiario",
                "concepto",
                "referencia"
            ]
        }
    },
    temperature=0
)


print(respuesta.choices[0].message.content)
```
