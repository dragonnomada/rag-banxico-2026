# Contexto FAISS con LLM

```python
from sentence_transformers import SentenceTransformer

modelo = SentenceTransformer(
    "../modelos/embeddings/paraphrase-multilingual-MiniLM-L12-v2"
)
```

```python
from pathlib import Path
import json

import faiss
import numpy
```

```python
FAISS_DIR = Path(
    "../shared/output/financiera/faiss"
)

INDEX_FILE = FAISS_DIR / "index.faiss"
METADATA_FILE = FAISS_DIR / "metadata.json"

index = faiss.read_index(
    str(INDEX_FILE)
)

metadata = json.loads(
    METADATA_FILE.read_text(
        encoding="utf-8"
    )
)
```

```python
pregunta = (
    "¿Cuál contiene Pago 16?"
)

query = modelo.encode(
    [pregunta],
    convert_to_numpy=True
)

query = numpy.asarray(
    query,
    dtype="float32"
)

faiss.normalize_L2(
    query
)

k = 5

scores, indices = index.search(
    query,
    k
)
```

```python
contexto = "\n\n".join(
    metadata[i]["texto"]
    for i in indices[0]
)
```

```python
from openai import OpenAI

cliente = OpenAI(
    base_url="http://localhost:8080/v1",
    api_key="local"
)
```

```python
respuesta = cliente.chat.completions.create(
    model="local",
    messages=[
        {
            "role": "system",
            "content": (
                "Responde únicamente usando "
                "el contexto proporcionado. "
                "Si la respuesta no está en "
                "el contexto, indícalo."
            )
        },
        {
            "role": "user",
            "content": f"""
                CONTEXTO:

                {contexto}

                PREGUNTA:

                {pregunta}
            """
        }
    ],
    temperature=0
)

print(respuesta.choices[0].message.content)
```
