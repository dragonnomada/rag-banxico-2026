# Consumo FAISS sin LLM

```python
from sentence_transformers import SentenceTransformer

modelo = SentenceTransformer(
    "../modelos/embeddings/paraphrase-multilingual-MiniLM-L12-v2"
)
```

```python
from pathlib import Path
import json

import faiss
import numpy
```

```python
FAISS_DIR = Path(
    "../shared/output/financiera/faiss"
)

INDEX_FILE = FAISS_DIR / "index.faiss"
METADATA_FILE = FAISS_DIR / "metadata.json"

index = faiss.read_index(
    str(INDEX_FILE)
)

metadata = json.loads(
    METADATA_FILE.read_text(
        encoding="utf-8"
    )
)
```

```python
pregunta = (
    "¿Cuáles son las condiciones "
    "de pago del contrato?"
)

query = modelo.encode(
    [pregunta],
    convert_to_numpy=True
)

query = numpy.asarray(
    query,
    dtype="float32"
)

faiss.normalize_L2(
    query
)

k = 5

scores, indices = index.search(
    query,
    k
)

for score, i in zip(
    scores[0],
    indices[0]
):
    item = metadata[i]

    print("=" * 80)
    print(
        f"Score: {score:.4f}"
    )
    print(
        f"Tipo: {item['tipo']}"
    )
    print(
        f"Contrato: {item['contrato']}"
    )
    print(
        f"Archivo: {item['archivo']}"
    )
    print("-" * 80)
    print(
        item["texto"]
    )
```
